```python
# use BERT
import transformers 
transformers.TFBertTokenizer
tokenizer = transformers.TFBertTokenizer.from_pretrained(
    "bert-base-uncased",
    return_attention_mask=False,
    return_token_type_ids=False
)
tokenizer(['hello world'])
```

```python
from collections import OrderedDict

# Create the ordered dictionary
d = OrderedDict([('ana', 3), ('maria', 5), ('ion', 9)])

# Iterate over the items and find the key with value 5
for key, value in d.items():
    if value == 5:
        desired_key = key
        break

print(desired_key)
print(len(d))
```
